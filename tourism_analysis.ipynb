```python
# ==========================================
# VERIFY TABLE RELATIONSHIPS
# ==========================================

def check_relationship(parent_df, parent_col, child_df, child_col, relationship_name):
    
    # Parent table ke IDs
    parent_ids = set(parent_df[parent_col].dropna())

    # Child table ke IDs
    child_ids = set(child_df[child_col].dropna())

    # Child mein jo IDs parent mein nahi mil rahe
    unmatched = child_ids - parent_ids

    print("\n" + "=" * 60)
    print(relationship_name)
    print("=" * 60)

    print("Parent unique IDs :", len(parent_ids))
    print("Child unique IDs  :", len(child_ids))
    print("Unmatched IDs     :", len(unmatched))

    if len(unmatched) == 0:
        print("✅ Relationship VALID")
    else:
        print("⚠️ Relationship has unmatched IDs")
        print("Sample unmatched IDs:", list(unmatched)[:10])


# ==========================================
# TRANSACTION RELATIONSHIPS
# ==========================================

check_relationship(
    df["User"],
    "UserId",
    df["Transaction"],
    "UserId",
    "Transaction.UserId → User.UserId"
)

check_relationship(
    df["Item"],
    "AttractionId",
    df["Transaction"],
    "AttractionId",
    "Transaction.AttractionId → Item.AttractionId"
)

check_relationship(
    df["Mode"],
    "VisitModeId",
    df["Transaction"],
    "VisitMode",
    "Transaction.VisitMode → Mode.VisitModeId"
)


# ==========================================
# USER RELATIONSHIPS
# ==========================================

check_relationship(
    df["Continent"],
    "ContinentId",
    df["User"],
    "ContinentId",
    "User.ContinentId → Continent.ContinentId"
)

check_relationship(
    df["Region"],
    "RegionId",
    df["User"],
    "RegionId",
    "User.RegionId → Region.RegionId"
)

check_relationship(
    df["Country"],
    "CountryId",
    df["User"],
    "CountryId",
    "User.CountryId → Country.CountryId"
)

check_relationship(
    df["City"],
    "CityId",
    df["User"],
    "CityId",
    "User.CityId → City.CityId"
)


# ==========================================
# COUNTRY → REGION
# ==========================================

check_relationship(
    df["Region"],
    "RegionId",
    df["Country"],
    "RegionId",
    "Country.RegionId → Region.RegionId"
)


# ==========================================
# REGION → CONTINENT
# ==========================================

check_relationship(
    df["Continent"],
    "ContinentId",
    df["Region"],
    "ContinentId",
    "Region.ContinentId → Continent.ContinentId"
)


# ==========================================
# ITEM RELATIONSHIPS
# ==========================================

check_relationship(
    df["City"],
    "CityId",
    df["Item"],
    "AttractionCityId",
    "Item.AttractionCityId → City.CityId"
)

check_relationship(
    df["Type"],
    "AttractionTypeId",
    df["Item"],
    "AttractionTypeId",
    "Item.AttractionTypeId → Type.AttractionTypeId"
)
```

```python
# Original AttractionTypeId ko safe rakho
df["Item"]["OriginalAttractionTypeId"] = df["Item"]["AttractionTypeId"]


# Attraction category create karna
def get_attraction_category(row):
    value = row["AttractionTypeId"]

    # Agar value text hai
    if isinstance(value, str):
        return value

    # Agar value numeric ID hai
    type_match = df["Type"][
        df["Type"]["AttractionTypeId"] == value
    ]

    if not type_match.empty:
        return type_match.iloc[0]["AttractionType"]

    return "Unknown"


df["Item"]["AttractionCategory"] = df["Item"].apply(
    get_attraction_category,
    axis=1
)
```

```python
# Actual numerical columns
numeric_features = [
    "VisitYear",
    "VisitMonth",
    "Rating"
]

for col in numeric_features:
    Q1 = master[col].quantile(0.25)
    Q3 = master[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = master[
        (master[col] < lower_bound) |
        (master[col] > upper_bound)
    ]

    print(f"\n{col}")
    print("-" * 40)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Outliers:", len(outliers))
```

Outlier Analysis: IQR-based analysis identified 3,298 low-rating observations as statistical outliers. However, ratings from 1 to 5 are valid values in the dataset, so these observations were retained to preserve genuine user feedback.

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

master["Season"] = master["VisitMonth"].apply(get_season)
```
